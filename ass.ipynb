```python
df= sns.load_dataset("taxis")
```

```python

num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())


cat_cols = df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert pickup column to datetime
df['pickup'] = pd.to_datetime(df['pickup'])

# Line Chart
plt.figure(figsize=(10, 5))
plt.plot(df['pickup'], df['fare'])

plt.xlabel('Pickup Time')
plt.ylabel('Fare')
plt.title('Fare Over Time')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count trips for each payment method
payment_counts = df['payment_method'].value_counts()

# Pie Chart
plt.figure(figsize=(7, 7))
plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct='%1.1f%%'
)

plt.title('Distribution of Trips by Payment Method')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Histogram
plt.figure(figsize=(8, 5))

plt.hist(df['distance'], bins=20)

plt.xlabel('Distance')
plt.ylabel('Number of Trips')
plt.title('Distribution of Trip Distance')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Box Plot
df.boxplot(column='tip', by='pickup_borough', figsize=(8, 5))

plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')
plt.title('Distribution of Tip Amounts by Pickup Borough')

plt.suptitle('')  # Remove default title
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Count Plot
sns.countplot(data=df, x='pickup_borough')

plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')
plt.title('Number of Trips by Pickup Borough')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter Plot
sns.scatterplot(
    data=df,
    x='distance',
    y='fare',
    hue='pickup_borough'
)

plt.xlabel('Distance')
plt.ylabel('Fare')
plt.title('Relationship Between Distance and Fare')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
numeric_data = df[['distance', 'fare', 'tip', 'tolls', 'total']]

# Create correlation matrix
correlation_matrix = numeric_data.corr()

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Pair Plot
sns.pairplot(
    df,
    vars=['distance', 'fare', 'tip', 'total'],
    hue='pickup_zone'
)

plt.suptitle('Pairwise Relationships of Distance, Fare, Tip and Total',
             y=1.02)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Violin Plot
sns.violinplot(
    data=df,
    x='payment_method',
    y='fare'
)

plt.xlabel('Payment Method')
plt.ylabel('Fare')
plt.title('Distribution of Fare by Payment Method')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
